````python
import numpy as np
import pandas as pd
````

````python
data = {
    "date":
    ["2025-01-01",
        "2025-01-02",
        "2025-02-03",
        "2025-01-04",
        "2025-01-05"
    ],
    "temperature": [10, 15, 20, 25, 30]
}
````

````python
# Change to dataframe
df = pd.DataFrame(data)
df
````

````python
df.dtypes
````

````python
# Change to datetime
df.date = pd.to_datetime(df.date)
df.dtypes
````

````python
df.date.dt.day_name()
````

````python
df.date.dt.year
````

````python
df.date.dt.day
````

````python
df["month"] = df.date.dt.month # can't use df.month = df.date.dt.month , can't use attribute name to create a column
df
````

````python
df = df.set_index("date")
df
````

````python
print(df.index)
````

````python
df.loc["2025-02-03"]
````

````python
# Creating a bunch of dates
dates1 = pd.date_range("2026-05-01", "2026-05-5")
print(dates1)
````

````python
dates2= pd.date_range("2026-05-01", periods=10)
print(dates2)
````

````python
from datetime import date
# hourly
dates3 = pd.date_range(date(2026,5,1), periods=10, freq="h")
print(dates3)
````

````python
# every 15 mins
dates4 = pd.date_range("2025-05-04", periods = 10, freq = "15 min")
print(dates4)
````

````python
# Sorting index
df = df.sort_index()
df
````

````python
# print day of the week
df.index.day_name()
````

````python
# Print Month
df.index.month
````

````python
# Print Day
df.index.day
````

````python
# print year
df.index.year
````
